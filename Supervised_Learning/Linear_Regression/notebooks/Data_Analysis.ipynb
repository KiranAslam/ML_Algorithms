```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns

df_1 = pd.read_csv('../Data/Linear_success.csv')
df_2 = pd.read_csv('../Data/Linear_failure.csv')

#print(df_1.head())
#print(df_1.describe())
#print(df_2.head())
#print(df_2.describe())

fig, axes= plt.subplots(1,2 , figsize=(12,6), sharey=True)
sns.scatterplot(ax=axes[0], data=df_1, x='x', y='y', color='blue', s=100, label='Actual Data')
axes[0].set_title('Dataset I: Ideal Linear Trend\n(Success Case)', fontsize=14)
axes[0].set_xlabel('Independent Variable (x)', fontsize=12)
axes[0].set_ylabel('Dependent Variable (y)', fontsize=12)
axes[0].grid(True,linestyle='--', alpha=0.6)

sns.scatterplot(ax=axes[1], data=df_2, x='x', y='y', color='red', s=100, label='Actual Data')
axes[1].set_title('Dataset II: Deviation from Linear Trend\n(Failure Case)', fontsize=14)
axes[1].set_xlabel('Independent Variable (x)', fontsize=12)
axes[1].grid(True,linestyle='--', alpha=0.6)

plt.suptitle("Why Visualization Matters: Anscombe's Quartet Comparison", fontsize=16, fontweight='bold')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()


plt.figure(figsize=(12, 6))
sns.kdeplot(df_1['y'], fill=True, label='Dataset I (Success)', color='blue', alpha=0.5)
sns.kdeplot(df_2['y'], fill=True, label='Dataset II (Failure)', color='red', alpha=0.5)

plt.title("Density Distribution of Y-Values: Identical Spread!", fontsize=14)
plt.xlabel("Value of Y")
plt.ylabel("Density")
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=True)
sns.regplot(ax=axes[0], data=df_1, x='x', y='y', color='blue', scatter_kws={'s':100}, line_kws={'color':'black'})
axes[0].set_title('Dataset I: Line Follows the Trend\n(Appropriate Fit)', fontsize=14)
axes[0].grid(True, linestyle='--', alpha=0.6)

sns.regplot(ax=axes[1], data=df_2, x='x', y='y', color='red', scatter_kws={'s':100}, line_kws={'color':'black'})
axes[1].set_title('Dataset II: Line Misses the Curve\n(Underfitting)', fontsize=14)
axes[1].grid(True, linestyle='--', alpha=0.6)

plt.suptitle("Linear Regression Line on Success vs. Failure Data", fontsize=16, fontweight='bold')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()


res_1 = df_1['y'] - (0.5 * df_1['x'] + 3.0)
res_2 = df_2['y'] - (0.5 * df_2['x'] + 3.0)

fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

sns.scatterplot(ax=axes[0], x=df_1['x'], y=res_1, color='blue', s=100)
axes[0].axhline(y=0, color='black', linestyle='--')
axes[0].set_title('Dataset I Residuals:\nRandom (Good Fit)', fontsize=14)
axes[0].set_ylabel('Residual (Error)')

sns.scatterplot(ax=axes[1], x=df_2['x'], y=res_2, color='red', s=100)
axes[1].axhline(y=0, color='black', linestyle='--')
axes[1].set_title('Dataset II Residuals:\nNon-Random Pattern (Bad Fit)', fontsize=14)

plt.suptitle("Residual Analysis: The Ultimate Test", fontsize=16, fontweight='bold')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
