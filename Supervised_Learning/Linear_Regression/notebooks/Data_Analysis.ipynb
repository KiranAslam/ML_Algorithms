```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df_1 = pd.read_csv('../Data/Linear_success.csv')
df_2 = pd.read_csv('../Data/Linear_failure.csv')
df_3 = pd.read_csv('../Data/The_Outlier_Trick.csv')
df_4 = pd.read_csv('../Data/The-Vertical_Trap.csv')

fig, axes = plt.subplots(2, 2, figsize=(15, 11), sharex=True, sharey=True)
axes = axes.flatten() 
datasets = [df_1, df_2, df_3, df_4]
colors = ['blue', 'red', 'green', 'purple']
titles = [
    'Dataset I: Clean Linear Trend\n(Ideal Success Fit)',
    'Dataset II: Non-Linear Curve\n(Underfitting / Bad Fit)',
    'Dataset III: The Outlier Trick\n(Extreme Sensitivity)',
    'Dataset IV: The Vertical Trap\n(High Leverage Influence)'
]

for i, df_sub in enumerate(datasets):
   
    sns.regplot(ax=axes[i], data=df_sub, x='x', y='y', color=colors[i], 
                scatter_kws={'s':100}, line_kws={'color':'black', 'linewidth':2})
    
    axes[i].set_title(titles[i], fontsize=13, fontweight='bold', pad=10)
    axes[i].set_xlabel('X Variable', fontsize=11)
    axes[i].set_ylabel('Y Variable', fontsize=11)
    axes[i].grid(True, linestyle='--', alpha=0.5)

plt.suptitle("The Complete Anscombe's Quartet: How Linear Regression Fails Geometrically", 
             fontsize=18, fontweight='bold', y=0.98)

plt.tight_layout()
plt.show()
```
